```python
import tensorflow as tf
```

```python
import glob
import imageio
import matplotlib.pyplot as plt
import numpy as np
import os
import PIL
from tensorflow.keras import layers
import time
from tqdm import tqdm
from IPython import display
```

```python
(train_images, train_labels), _ = tf.keras.datasets.mnist.load_data()
```

```python
train_images = train_images.reshape(train_images.shape[0], 28, 28, 1).astype('float32')
train_images = (train_images -127.5)/127.5
len(train_images)
```

```python
BUFFER_SIZE = 60000
BATCH_SIZE = 256
```

```python

# One-hot encode labels (digits 0-9)
train_labels = tf.keras.utils.to_categorical(train_labels, 10)

train_dataset = tf.data.Dataset.from_tensor_slices((train_images, train_labels))
train_dataset = train_dataset.shuffle(BUFFER_SIZE)
train_dataset = train_dataset.batch(BATCH_SIZE)
```

```python
def generator_model():
    noise_input = layers.Input(shape=(100,))
    label_input = layers.Input(shape=(10,))
    
    # Concatenate noise and label vectors
    merged_input = layers.Concatenate()([noise_input, label_input])
    
    x = layers.Dense(7*7*256, use_bias=False)(merged_input)
    x = layers.BatchNormalization()(x)
    x = layers.LeakyReLU()(x)

    x = layers.Reshape((7, 7, 256))(x)

    x = layers.Conv2DTranspose(128, (5,5), strides=(1,1), padding='same', use_bias=False)(x)
    x = layers.BatchNormalization()(x)
    x = layers.LeakyReLU()(x)

    x = layers.Conv2DTranspose(64, (5,5), strides=(2,2), padding='same', use_bias=False)(x)
    x = layers.BatchNormalization()(x)
    x = layers.LeakyReLU()(x)

    x = layers.Conv2DTranspose(1, (5,5), strides=(2,2), padding='same', use_bias=False, activation='tanh')(x)

    return tf.keras.Model(inputs=[noise_input, label_input], outputs=x, name="cgan_generator")
```

```python
generator = generator_model()
noise = tf.random.normal([1, 100])

# Fix: Wrap the categorical array in tf.constant
label = tf.constant(tf.keras.utils.to_categorical([7], 10), dtype=tf.float32)

generated_image = generator([noise, label], training=False)

plt.imshow(generated_image[0, :, :, 0] * 127.5 + 127.5, cmap='gray')
plt.axis('off')
plt.show()
```

```python
generator.summary()
```

```python
def discriminator_model():
    img_input = layers.Input(shape=(28, 28, 1))
    label_input = layers.Input(shape=(10,))
    
    # Project label to match image spatial dimensions (28x28x1)
    label_embedding = layers.Dense(28 * 28)(label_input)
    label_embedding = layers.Reshape((28, 28, 1))(label_embedding)
    
    # Concatenate image and label channels -> (None, 28, 28, 2)
    x = layers.Concatenate()([img_input, label_embedding])

    x = layers.Conv2D(64, (5,5), strides=(2,2), padding='same')(x)
    x = layers.LeakyReLU()(x)
    x = layers.Dropout(0.3)(x)

    x = layers.Conv2D(128, (5,5), strides=(2,2), padding='same')(x)
    x = layers.LeakyReLU()(x)
    x = layers.Dropout(0.3)(x)

    x = layers.Flatten()(x)
    validity = layers.Dense(1, activation="sigmoid")(x)

    return tf.keras.Model(inputs=[img_input, label_input], outputs=validity, name="cgan_discriminator")
```

```python
discriminator = discriminator_model()

# Create a matching label tensor for what the image is supposed to be (e.g., digit 7)
test_label = tf.constant(tf.keras.utils.to_categorical([7], 10), dtype=tf.float32)

# Pass both the image and the label inside a list
decision = discriminator([generated_image, test_label])
print(decision)
```

```python
discriminator.summary()
```

```python
cross_entropy = tf.keras.losses.BinaryCrossentropy(from_logits = True)
```

```python
def discriminator_loss(real_output, fake_output):
    real_loss = cross_entropy(tf.ones_like(real_output), real_output)
    fake_loss = cross_entropy(tf.zeros_like(fake_output), fake_output)
    total_loss = real_loss + fake_loss
    return total_loss
```

```python
def generator_loss(fake_output):
    return cross_entropy(tf.ones_like(fake_output), fake_output)
```

```python
generator_optimizer  = tf.keras.optimizers.Adam(1e-4)
discriminator_optimizer = tf.keras.optimizers.Adam(1e-4)
```

```python
checkpoint_dir = './training_checkpoints'
checkpoint_prefix = os.path.join(checkpoint_dir, "ckpt")
checkpoint = tf.train.Checkpoint(generator_optimizer = generator_optimizer, discriminator_optimizer = discriminator_optimizer,
                                 generator = generator, discriminator = discriminator)
```

```python
EPOCHS = 50
noise_dim = 100
num_examples_to_generate = 16
seed = tf.random.normal([num_examples_to_generate, noise_dim])
```

```python
@tf.function
def train_steps(images, labels):
    # Dynamically grab the actual batch size from the current batch of images
    current_batch_size = tf.shape(images)[0]
    
    # Use current_batch_size instead of BATCH_SIZE
    noise = tf.random.normal([current_batch_size, noise_dim])

    with tf.GradientTape() as gen_tape, tf.GradientTape() as disc_tape:
        generated_images = generator([noise, labels], training=True)

        real_output = discriminator([images, labels], training=True)
        fake_output = discriminator([generated_images, labels], training=True)

        gen_loss = generator_loss(fake_output)
        disc_loss = discriminator_loss(real_output, fake_output)

    gradients_of_generator = gen_tape.gradient(gen_loss, generator.trainable_variables)
    gradients_of_discriminator = disc_tape.gradient(disc_loss, discriminator.trainable_variables)

    generator_optimizer.apply_gradients(zip(gradients_of_generator, generator.trainable_variables))
    discriminator_optimizer.apply_gradients(zip(gradients_of_discriminator, discriminator.trainable_variables))
```

```python
def train(dataset, epochs):
    for epoch in range(epochs):
        start = time.time()

        for image_batch, label_batch in tqdm(dataset):
            train_steps(image_batch, label_batch)

        display.clear_output(wait = True)
        generate_and_save_images(generator, epoch+1, seed)

        if (epoch+1) % 15 == 0:
            checkpoint.save(file_prefix = checkpoint_prefix)

        print ('Time for epoch {} is {} sec'.format(epoch+1, time.time()-start))

    display.clear_output(wait= True)
    generate_and_save_images(generator, epochs, seed)
```

```python
EPOCHS = 50
noise_dim = 100
num_examples_to_generate = 16

# Create fixed noise and fixed labels for consistent evaluation grids
seed_noise = tf.random.normal([num_examples_to_generate, noise_dim])

# Fix: Wrap the categorical array in tf.constant
seed_labels = tf.constant(
    tf.keras.utils.to_categorical([i % 10 for i in range(num_examples_to_generate)], 10), 
    dtype=tf.float32
)

seed = (seed_noise, seed_labels)

def generate_and_save_images(model, epoch, test_input):
    noise_input, label_input = test_input
    predictions = model([noise_input, label_input], training=False)
    fig = plt.figure(figsize = (4,4))

    for i in range(predictions.shape[0]):
        plt.subplot(4,4,i+1)
        plt.imshow(predictions[i, :, :, 0] * 127.5 + 127.5, cmap = 'gray')
        plt.axis('off')

    plt.savefig('Image_at_epoch{:04d}.png'.format(epoch))
    plt.show()
```

```python
train(train_dataset,EPOCHS)
```
